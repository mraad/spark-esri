```python
import os
import warnings

from spark_esri import spark_start, spark_stop

from sparkgeo import register_st_functions

warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
os.path.join(os.environ['SG_HOME'],"sparkgeo-3.2.1-0.16.zip")
```

```python
extra_java_options = "-XX:+UseCompressedOops -XX:+AggressiveHeap"

config = {
    "spark.driver.memory": "31G",
    "spark.executor.memory": "31G",
    "spark.driver.extraJavaOptions": extra_java_options,
    "spark.executor.extraJavaOptions": extra_java_options,
    "spark.serializer": "org.apache.spark.serializer.KryoSerializer",
    "spark.kryo.registrator": "com.esri.spark.Registrator",
    "spark.kryo.unsafe": True,
    "spark.memory.offHeap.enable": True,
    "spark.memory.offHeap.size": "31G",
    "spark.jars": os.path.join(os.environ['SG_HOME'],"sparkgeo-3.2.1-0.16.jar"),
    "spark.submit.pyFiles": os.path.join(os.environ['SG_HOME'],"sparkgeo-3.2.1-0.16.zip")
}

spark_stop()
spark = spark_start(config=config)
```

```python
register_st_functions()
```

```python
spark\
    .range(1_000_000)\
    .selectExpr("id","-180.0+360.0*rand() lon","-90.0+180.0*rand() lat")\
    .selectExpr("id","lonToX(lon) x","latToY(lat) y")\
    .selectExpr("avg(x) avg_x","avg(y) avg_y")\
    .toPandas()
```
