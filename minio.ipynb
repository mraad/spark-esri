```python
import os
import pyarrow.parquet as pq

from pyarrow import fs
from s3fs import S3FileSystem
```

```python
file_sys = S3FileSystem(anon=False,
                       key="minioadmin",
                       secret="minioadmin",
                       client_kwargs={'endpoint_url':'http://localhost:9000'}
                 )
```

```python
base_path = "output/pip.par"
base_name = f"{base_path}/part-"
```

```python
parts = list(filter(lambda f:f.startswith(base_name), file_sys.ls("output/pip.par")))
```

```python
for p in parts:
    print(p)
```

```python
table = pq.read_table(parts[0],filesystem=file_sys)
pydict = table.to_pydict()
```

```python
pydict.keys()
```

```python
file_sys = fs.LocalFileSystem()
```

```python
file_path = os.path.join("Z:",os.sep,"PGE","output","pip.par")

parts = file_sys.get_file_info(fs.FileSelector(file_path, recursive=True))
parts = filter(lambda f:f.base_name.startswith("part-"), parts)
parts = list(map(lambda f:f.path, parts))
```

```python
for f in parts:
    print(f)
```

```python
table = pq.read_table(parts[0],filesystem=file_sys)
pydict = table.to_pydict()
```
