```python
import os
from spark_esri import spark_start, spark_stop
```

```python
spark_stop()

config = {
    "spark.driver.memory":"16G",
    "spark.executor.memory":"16G",
    "spark.sql.execution.arrow.pyspark.enabled":True
}
spark = spark_start(config=config)
```

```python
spark.version
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)


schema = StructType(
    [
        StructField("ID", IntegerType()),
        StructField("COMMODITY", StringType()),
        StructField("SITE_NAME", StringType()),
        StructField("COMPANY_NA", StringType()),
        StructField("STATE_LOCA", StringType()),
        StructField("COUNTY", StringType()),
        StructField("LATITUDE", DoubleType()),
        StructField("LONGITUDE", DoubleType()),
        StructField("PLANT_MIN", StringType()),
    ]
)

path = os.path.join("Z:",os.sep,"GWorkspace","spark_esri","mineplant_2.txt")
mines = spark.read.csv(path=path, schema=schema, sep="\t", header=True)
mines.printSchema()
mines.show(3)
mines.count()
```

```python
pd = mines.toPandas()
pd
```

```python
path = os.path.join("C:",os.sep,"TEMP","mines.prq")
mines\
    .write\
    .parquet(path,mode="overwrite")
```
