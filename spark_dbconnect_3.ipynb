```python
import os
import json

from esri_spark import register_functions

from spark_dbconnect import spark_start, spark_stop
```

```python
filename = os.path.expanduser(os.path.join("~",".databricks-connect"))
with open(filename) as f:
  db_connect = json.load(f)
```

```python
esri_spark_jar = os.path.join("C:",os.sep,"jars","esri_spark_2.4.7-0.20.jar")

config = {
    "spark.databricks.service.server.enabled": True,
    "spark.databricks.service.client.enabled": True,
    "spark.databricks.service.address": db_connect['host'],
    "spark.databricks.service.token": db_connect['token'],
    "spark.databricks.service.clusterId": db_connect['cluster_id'],
    "spark.databricks.service.orgId": db_connect['org_id'],
    "spark.databricks.service.port": db_connect['port'],
    "spark.driver.extraClassPath": esri_spark_jar,
    "spark.sql.catalogImplementation": "hive",
    "spark.sql.extensions": "dbfs:/esri/esri_spark_2.4.7-0.20.jar",
    "spark.esri.license.path": "dbfs:/esri/mraad.lic"
}

spark = spark_start(config=config)
```

```python
spark.conf.get("spark.master","Undefined")
```

```python
register_functions("dbfs:/esri/mraad.lic")
```
